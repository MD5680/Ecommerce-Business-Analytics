####Relationship Validation

```python
%sql
SELECT
    COUNT(*) AS unmatched_orders
FROM raw_orders o
LEFT JOIN raw_customers c
    ON o.customer_id = c.customer_id
WHERE c.customer_id IS NULL;
```

```python
%sql
SELECT
    COUNT(*) AS unmatched_order_items
FROM raw_order_items oi
LEFT JOIN raw_orders o
    ON oi.order_id = o.order_id
WHERE o.order_id IS NULL;
```

```python
%sql
SELECT
    COUNT(*) AS unmatched_products
FROM raw_order_items oi
LEFT JOIN raw_products p
    ON oi.product_id = p.product_id
WHERE p.product_id IS NULL;
```

```python
%sql
SELECT
    COUNT(*) AS unmatched_payments
FROM raw_order_payments op
LEFT JOIN raw_orders o
    ON op.order_id = o.order_id
WHERE o.order_id IS NULL;
```
