```python
customers_df = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv("/Volumes/workspace/default/ecommerce-datasets/e_commerce_datasets/olist_customers_dataset.csv")
```

```python
customers_df.display()
```

##Check the schema

```python
customers_df.printSchema()
```

###Check row count

```python
customers_df.count()
```

###Save as a Delta table

```python
customers_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("raw_customers")
```

###Verify using SQL

```python
%sql

SELECT *
FROM raw_customers
LIMIT 10;
```

```python
%sql
select count(*) as total_records from raw_customers
```
